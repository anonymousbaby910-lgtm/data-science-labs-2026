```python
# Example: calculate the average value
# for each row in a dataset
import pandas as pd

data = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78}
]

df = pd.DataFrame(data)

# Loop through each row
for index, row in df.iterrows():
    name = row["name"]
    score = row["score"]
    print(f"{name}: {score}")
```

```python
scores = [55, 88, 72, 40, 95]
passed = []

for s in scores:
    if s >= 60:
        passed.append(s)
    else:
        print(f"Review needed: {s}")

pass_rate = len(passed) / len(scores)

if pass_rate >= 0.8:
    print("Strong cohort result")
elif pass_rate >= 0.5:
    print("Mixed result")
else:
    print("Needs review")
```

```python
def normalize(values):
    """Scale values to a 0-1 range."""
    low = min(values)
    high = max(values)
    return [(v - low) / (high - low)
            for v in values]

def average(values):
    return sum(values) / len(values)

grades = [88, 91, 76, 95, 84]

scaled = normalize(grades)
mean_grade = average(grades)

print(mean_grade)
print(scaled)
```

```python
# Writing lines to a file
students = ["Aiyana", "Minho", "Zainab"]

with open("roster.txt", "w") as f:
    for name in students:
        f.write(name + "\n")

# Reading the file back
with open("roster.txt", "r") as f:
    lines = f.readlines()

names = [line.strip() for line in lines]
print(names)

# ['Aiyana', 'Minho', 'Zainab']
```

```python
temps_celsius = np.array([
    [20, 22, 19],
    [25, 27, 24]
])

# A single number broadcasts
# across the whole array
temps_fahrenheit = temps_celsius * 9/5 + 32

# A 1D array broadcasts across rows
row_adjustment = np.array([1, 0, -1])
adjusted = temps_celsius + row_adjustment

print(adjusted)
# [[21 22 18]
#  [26 27 23]]
```

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5], dtype=float)
b = np.array([10, 20, 30, 40, 50], dtype=float)

result = []
for x, y in zip(a, b):
    result.append(x + y)

result = np.array(result)
```

```python
import numpy as np
import time

n = 5_000_000
a = list(range(n))
b = np.arange(n)

start = time.time()
result = [x * 2 for x in a]
print("Python loop:", time.time() - start)

start = time.time()
result = b * 2
print("NumPy vectorized:", time.time() - start)

# Python loop:      ~0.35 sec
# NumPy vectorized: ~0.01 sec
```
